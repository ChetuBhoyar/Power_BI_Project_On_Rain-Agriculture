#  Data Cleaning

Based on the initial data profiling, we'll now clean the dataset by handling missing values, standardizing categorical columns, and addressing potential outliers.

```python
# Create a copy of the original DataFrame for cleaning
df_clean = df.copy()
```

###  Standardize Categorical Data

We'll standardize the 'make' column to a consistent case and address the '—' entry in the 'color' column, treating it as a missing value.

```python
# Standardize 'make' column to lowercase for consistency
df_clean['make'] = df_clean['make'].str.lower()

# Standardize 'color' column: replace '—' with NaN and then convert to lowercase
df_clean['color'] = df_clean['color'].replace('—', np.nan)
df_clean['color'] = df_clean['color'].str.lower()

# Standardize 'interior' column: replace '—' with NaN and then convert to lowercase
df_clean['interior'] = df_clean['interior'].replace('—', np.nan)
df_clean['interior'] = df_clean['interior'].str.lower()

print("Categorical columns 'make', 'color', and 'interior' standardized.")
print("Unique makes (sample):")
print(df_clean['make'].value_counts().head())
print("Unique colors (sample):")
print(df_clean['color'].value_counts().head())
print("Unique interiors (sample):")
print(df_clean['interior'].value_counts().head())
```

###  Handle Missing Values

We'll address missing values in various columns:
- Drop rows where `sellingprice`, `mmr`, or `seller` are missing (very few).
- Impute `condition` and `odometer` (after outlier treatment) with their respective medians.
- Fill categorical missing values in `make`, `model`, `trim`, `body`, `transmission`, `color`, and `interior` with 'Unknown'.

```python
# Drop rows with missing values in critical columns: sellingprice, mmr, seller
df_clean = df_clean.dropna(subset=['sellingprice', 'mmr', 'seller'])
print(f"DataFrame shape after dropping rows with missing sellingprice, mmr, or seller: {df_clean.shape}")

# Impute numerical columns with median
# For 'condition', fill missing values with the median
df_clean['condition'] = df_clean['condition'].fillna(df_clean['condition'].median())
print(f"Missing values in 'condition' after median imputation: {df_clean['condition'].isnull().sum()}")

# For 'odometer', first handle the outlier '999999.0' (treated as NaN for imputation)
df_clean['odometer'] = df_clean['odometer'].replace(999999.0, np.nan)
# Then, fill remaining missing 'odometer' values with the median
df_clean['odometer'] = df_clean['odometer'].fillna(df_clean['odometer'].median())
print(f"Missing values in 'odometer' after outlier replacement and median imputation: {df_clean['odometer'].isnull().sum()}")

# Impute categorical columns with 'Unknown'
for col in ['make', 'model', 'trim', 'body', 'transmission', 'color', 'interior']:
    df_clean[col] = df_clean[col].fillna('unknown')
print("Categorical missing values filled with 'unknown'.")

# Verify that there are no more missing values after cleaning
print("\n--- Null Values After Cleaning ---")
print(df_clean.isnull().sum())
```

###  Address Outliers (Odometer and Selling Price)

We've already handled the `999999.0` outlier in `odometer`. Now let's review `sellingprice` for potential outliers, specifically the `$1.00` entry previously identified, and consider its impact.

```python
# Check for selling prices that are extremely low, e.g., $1.00
extremely_low_price_cars = df_clean[df_clean['sellingprice'] <= 10]

print(f"Number of cars with selling price $10 or less: {len(extremely_low_price_cars)}")
display(extremely_low_price_cars)

# Decision: For now, we'll keep these low values as they might represent specific scenarios.
# If these values significantly skew future analyses, they can be removed or imputed.
```

3.7 Plot a box plot to show the distribution of car selling prices grouped by
color. Extract insights and if required, remove outliers and plot the graph
again.


```python
# 3.7 Box plot with Outlier Removal
Q1 = df_clean['sellingprice'].quantile(0.25)
Q3 = df_clean['sellingprice'].quantile(0.75)
IQR = Q3 - Q1

# Define clean boundaries excluding extreme luxury/exotic anomalies
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Filter out the outliers and drop any ambiguous color codes like '—'
df_filtered_color = df_clean[
    (df_clean['sellingprice'] >= lower_bound) &
    (df_clean['sellingprice'] <= upper_bound) &
    (df_clean['color'] != '—')
]

plt.figure(figsize=(15, 7))
sns.boxplot(x='color', y='sellingprice', data=df_filtered_color, palette='Set3')
plt.title("Distribution of Car Selling Prices Grouped by Color (Outliers Removed)", fontsize=14)
plt.xlabel("Car Exterior Color", fontsize=12)
plt.ylabel("Selling Price ($)", fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
